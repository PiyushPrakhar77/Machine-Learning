```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("/content/ecommerce_data.csv")
data.head()
```

```python
plt.figure(figsize=(6, 4))
plt.plot(data['month_number'], data['total_profit'], marker='o', linestyle='-', color='b', label='Total Profit')

plt.xlabel('Month')
plt.ylabel('Total Profit')
plt.title('Total Profit Month Wise')

plt.grid(True)
plt.legend(loc='upper left')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(data['month_number'], data['t shirts'], color='r', label='T-shirts')

plt.xlabel('Month')
plt.ylabel('T-shirts Sold')
plt.title('T-shirts Sales Month Wise')

plt.legend(loc='upper right')
plt.grid(True, linestyle='--', linewidth=1)
plt.show()
```

```python
import numpy as np

months = data['month_number']
trousers = data['trousers']
shirts = data['shirts']

bar_width = 0.3

x = np.arange(len(months))

plt.bar(x - bar_width/2, trousers, width=0.3, label='Trousers')
plt.bar(x + bar_width/2, shirts, width=0.3, label='Shirts')

plt.xlabel('Month')
plt.ylabel('Quantity Sold')
plt.title('Monthly Sales of Trousers and Shirts')

plt.legend(loc='upper left')
plt.grid(linewidth=1, linestyle='--')
plt.xticks(x, months)

plt.savefig('sales_chart.png', dpi=200)
plt.show()
```

```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
sns.displot(data=tips[['total_bill', 'tip']], kind='hist')
plt.suptitle("Histogram of total_bill & tip")
plt.show()
```

```python
sns.displot(data=tips[['total_bill', 'tip']], kind='kde')
plt.suptitle("KDE of total_bill & tip")
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title("Boxplot: total_bill by Day")
plt.show()
```

```python
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title("Violin Plot: total_bill by Day")
plt.show()
```

```python
corr = tips[['total_bill', 'tip', 'size']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
