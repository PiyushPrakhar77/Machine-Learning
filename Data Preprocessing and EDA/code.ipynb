```python
import pandas as pd
from sklearn.model_selection import train_test_split

try:
    train_df = pd.read_csv('titanic_train.csv')
    test_df = pd.read_csv('titanic_test.csv')
    merged_titanic_df = pd.concat([train_df, test_df], ignore_index=True)
    display(merged_titanic_df.head())
    print("\nShape of the merged dataframe:", merged_titanic_df.shape)
except FileNotFoundError:
    print("Error: One or both of the Titanic CSV files not found.")
```

```python
try:
    math_students_df = pd.read_csv('students_math.csv')
    portuguese_students_df = pd.read_csv('students_portuguese.csv')

    combined_students_df = pd.merge(math_students_df, portuguese_students_df, on='school')

    display(combined_students_df.head())
    print("\nShape of the combined dataframe:", combined_students_df.shape)

except FileNotFoundError:
    print("Error: One or both of the student data CSV files not found.")
except KeyError:
    print("Error: 'school' column not found in one or both dataframes.")
```
